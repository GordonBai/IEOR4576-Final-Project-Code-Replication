```python
# ============================================================
# MODULE 1B — Get LSEG futures contract families
# Lightweight version
# ============================================================

# Previous query killed the WRDS connection, reconnect first
db = wrds.Connection()

families = db.raw_sql("""
SELECT DISTINCT
    contrcode,
    clscode,
    contrname,
    exchtickersymb,
    ldb,
    isocurrcode
FROM tr_ds_fut.wrds_contract_info
WHERE contrcode IS NOT NULL
ORDER BY contrname
""")

print("Number of LSEG contract-family records:", len(families))

display(families.head(50))
```

```python
# ============================================================
# MODULE 1C — Candidate mapping:
# 64 paper markets -> LSEG contract families
# ============================================================

import re
from difflib import SequenceMatcher
import pandas as pd


def normalize_name(x):
    if pd.isna(x):
        return ""

    x = str(x).upper()

    replacements = {
        "&": " AND ",
        "#": " ",
        "-": " ",
        "/": " ",
        ".": " ",
        ",": " ",
        "FUTURES": " ",
        "FUTURE": " ",
        "INDEX": " ",
    }

    for a, b in replacements.items():
        x = x.replace(a, b)

    x = re.sub(r"\s+", " ", x).strip()
    return x


def similarity(a, b):
    a = normalize_name(a)
    b = normalize_name(b)

    if not a or not b:
        return 0

    # character similarity
    seq = SequenceMatcher(None, a, b).ratio()

    # token overlap
    ta = set(a.split())
    tb = set(b.split())

    overlap = len(ta & tb) / max(len(ta), 1)

    return 0.65 * seq + 0.35 * overlap


candidate_rows = []

for _, p in paper.iterrows():

    desc = p["paper_description"]
    ticker = p["paper_ticker"]

    temp = families.copy()

    temp["description_score"] = temp["contrname"].apply(
        lambda x: similarity(desc, x)
    )

    # Small bonus only — do NOT trust paper ticker blindly
    temp["ticker_bonus"] = (
        temp["exchtickersymb"]
        .astype(str)
        .str.upper()
        .eq(str(ticker).upper())
        .astype(float)
        * 0.15
    )

    temp["score"] = (
        temp["description_score"]
        + temp["ticker_bonus"]
    )

    temp = (
        temp.sort_values("score", ascending=False)
        .drop_duplicates(
            subset=[
                "contrcode",
                "clscode",
                "contrname",
                "exchtickersymb"
            ]
        )
        .head(8)
    )

    for rank, (_, r) in enumerate(temp.iterrows(), start=1):

        candidate_rows.append({
            "paper_ticker": ticker,
            "asset_class": p["asset_class"],
            "paper_description": desc,
            "paper_start": p["paper_start"],
            "paper_end": p["paper_end"],

            "rank": rank,
            "score": round(r["score"], 4),

            "contrcode": r["contrcode"],
            "clscode": r["clscode"],
            "lseg_name": r["contrname"],
            "lseg_ticker": r["exchtickersymb"],
            "ldb": r["ldb"],
            "currency": r["isocurrcode"],
        })


candidates = pd.DataFrame(candidate_rows)

print("Paper markets:", candidates["paper_ticker"].nunique())
print("Candidate rows:", len(candidates))

display(candidates.head(40))
```

```python
# ============================================================
# MODULE 1D — Re-rank LSEG family candidates
# using:
#   1. product-name similarity
#   2. paper qualifiers
#   3. historical coverage
#   4. ticker information
# ============================================================

import numpy as np
import pandas as pd
import re

# Load the candidate audit we already created earlier
old_candidates = pd.read_csv("paper64_lseg_candidates.csv")

# Add the corrected paper descriptions
old_candidates = old_candidates.merge(
    paper[
        ["paper_ticker", "paper_description"]
    ],
    on="paper_ticker",
    how="left"
)


# ------------------------------------------------------------
# Collapse families metadata to one row per contrcode
# ------------------------------------------------------------

def join_unique(x):
    vals = (
        pd.Series(x)
        .dropna()
        .astype(str)
        .unique()
    )
    return " | ".join(vals)


family_meta = (
    families
    .groupby("contrcode", as_index=False)
    .agg({
        "clscode": "first",
        "contrname": "first",
        "exchtickersymb": join_unique,
        "ldb": join_unique,
        "isocurrcode": join_unique,
    })
)

old_candidates = old_candidates.merge(
    family_meta,
    left_on="lseg_contrcode",
    right_on="contrcode",
    how="left"
)


# ------------------------------------------------------------
# Remove qualifiers from paper description
# so we can compare the underlying product name separately
# ------------------------------------------------------------

def base_description(x):

    x = str(x).upper()

    remove_terms = [
        "COMPOSITE",
        "ELECTRONIC",
        "DAY SESSION",
        "MINI",
        "CBOT",
    ]

    for term in remove_terms:
        x = x.replace(term, " ")

    x = x.replace("MINN", "MINNEAPOLIS")

    x = re.sub(r"[^A-Z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x).strip()

    return x


# ------------------------------------------------------------
# Qualifier match
# ------------------------------------------------------------

def qualifier_score(description, lseg_name):

    desc = str(description).upper()
    name = str(lseg_name).upper()

    score = 0

    if "COMPOSITE" in desc:
        if "COMP" in name or "COMPOSITE" in name:
            score += 2

    if "ELECTRONIC" in desc:
        if any(
            x in name
            for x in ["ELECTRONIC", "GLOBEX", "ECBOT"]
        ):
            score += 2

    if "MINI" in desc:
        if any(
            x in name
            for x in ["MINI", "EMINI", "E-MINI"]
        ):
            score += 2

    if "DAY SESSION" in desc:
        if "DAY SESSION" in name:
            score += 2

    if "CBOT" in desc:
        if any(x in name for x in ["CBOT", "CBT", "ECBOT"]):
            score += 1.5

    if "MINN" in desc:
        if "MINNEAPOLIS" in name:
            score += 2

    return score


# ------------------------------------------------------------
# Calculate review score
# ------------------------------------------------------------

old_candidates["base_name_score"] = old_candidates.apply(
    lambda r: similarity(
        base_description(r["paper_description"]),
        r["lseg_name"]
    ),
    axis=1
)

old_candidates["qualifier_score"] = old_candidates.apply(
    lambda r: qualifier_score(
        r["paper_description"],
        r["lseg_name"]
    ),
    axis=1
)

old_candidates["coverage_score"] = (
    old_candidates["date_coverage"]
    .eq("PASS")
    .astype(float)
    * 2
)

old_candidates["ticker_score"] = old_candidates.apply(
    lambda r:
        0.75
        if str(r["paper_ticker"]).upper()
        in str(r["exchtickersymb"]).upper().split(" | ")
        else 0,
    axis=1
)

old_candidates["rank_bonus"] = (
    0.5 / old_candidates["candidate_rank"]
)

old_candidates["review_score"] = (
    3 * old_candidates["base_name_score"]
    + old_candidates["qualifier_score"]
    + old_candidates["coverage_score"]
    + old_candidates["ticker_score"]
    + old_candidates["rank_bonus"]
)


# New rank within every paper market
old_candidates["review_rank"] = (
    old_candidates
    .groupby("paper_ticker")["review_score"]
    .rank(method="first", ascending=False)
)


review = (
    old_candidates[
        old_candidates["review_rank"] <= 3
    ]
    .sort_values(
        ["asset_class", "paper_ticker", "review_rank"]
    )
)


print("Paper markets:", review["paper_ticker"].nunique())
print("Rows:", len(review))

display(
    review[[
        "paper_ticker",
        "paper_description",
        "review_rank",
        "review_score",
        "lseg_contrcode",
        "lseg_name",
        "exchtickersymb",
        "lseg_earliest",
        "lseg_latest",
        "n_contracts",
        "date_coverage"
    ]]
)
```

```python
# ============================================================
# MODULE 1L — Build final paper-market → LSEG family map
# ============================================================

# ------------------------------------------------------------
# 1. Final primary family for each paper market
#    For stitched markets, primary = modern/latest family
# ------------------------------------------------------------

FINAL_UPDATES = {
    # partial-coverage fixes
    "AN": 2744,   # Australian Dollar Composite
    "BN": 2759,   # Sterling Composite
    "DT": 834,    # Euro Bund (LSEG starts 1998)
    "ER": 4396,   # CME E-mini Russell 2000 (modern leg)
    "MD": 401,    # E-mini S&P 400 electronic
    "YM": 444,    # Mini Dow
    "ZB": 2091,   # RBOB Gasoline
    "ZL": 3269,   # Soybean Oil Composite
    "ZR": 461,    # Rough Rice electronic
}

for ticker, contrcode in FINAL_UPDATES.items():
    mapping.loc[
        mapping["paper_ticker"] == ticker,
        "lseg_contrcode"
    ] = contrcode


# ------------------------------------------------------------
# 2. Families that require more than one historical segment
# ------------------------------------------------------------

CHAIN_OVERRIDES = {

    # Russell 2000
    # LSEG has a gap between the old index family and TF mini.
    "ER": [
        (1, 394,  "predecessor"),
        (2, 2101, "mini_TF"),
        (3, 4396, "mini_RTY"),
    ],

    # S&P MidCap 400:
    # composite full-size predecessor → electronic E-mini
    "MD": [
        (1, 3368, "predecessor_composite"),
        (2, 401,  "electronic_emini"),
    ],

    # Gasoline:
    # old NY unleaded gasoline → RBOB
    "ZB": [
        (1, 2032, "unleaded_predecessor"),
        (2, 2091, "RBOB"),
    ],

    # Rough Rice:
    # old RR family → electronic ZR
    "ZR": [
        (1, 379, "old_rough_rice"),
        (2, 461, "electronic_ZR"),
    ],
}


# ------------------------------------------------------------
# 3. Build one-row-per-family-segment table
# ------------------------------------------------------------

rows = []

for _, r in mapping.iterrows():

    ticker = r["paper_ticker"]

    if ticker in CHAIN_OVERRIDES:

        for order, code, role in CHAIN_OVERRIDES[ticker]:
            rows.append({
                "paper_ticker": ticker,
                "paper_description": r["paper_description"],
                "segment_order": order,
                "lseg_contrcode": code,
                "segment_role": role,
                "mapping_type": "stitched",
            })

    else:

        mapping_type = "single"

        if ticker in ["CL", "ZU", "NR", "ZN"]:
            mapping_type = "shared_family_proxy"

        if ticker == "DT":
            mapping_type = "partial_history"

        rows.append({
            "paper_ticker": ticker,
            "paper_description": r["paper_description"],
            "segment_order": 1,
            "lseg_contrcode": r["lseg_contrcode"],
            "segment_role": "main",
            "mapping_type": mapping_type,
        })


family_chain = pd.DataFrame(rows)


# ------------------------------------------------------------
# 4. Add readable LSEG family metadata
# ------------------------------------------------------------

family_lookup = (
    families[
        [
            "contrcode",
            "contrname",
            "exchtickersymb",
            "isocurrcode",
        ]
    ]
    .drop_duplicates("contrcode")
    .rename(columns={
        "contrcode": "lseg_contrcode",
        "contrname": "lseg_name",
        "exchtickersymb": "lseg_ticker",
        "isocurrcode": "currency",
    })
)

family_chain = (
    family_chain
    .merge(
        family_lookup,
        on="lseg_contrcode",
        how="left"
    )
    .sort_values(
        ["paper_ticker", "segment_order"]
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. QA
# ------------------------------------------------------------

print("Paper markets:", family_chain["paper_ticker"].nunique())
print("Family segments:", len(family_chain))
print()

print("Mapping types:")
display(
    family_chain[
        ["paper_ticker", "mapping_type"]
    ]
    .drop_duplicates()
    ["mapping_type"]
    .value_counts()
)

print("\nSpecial mappings:")
display(
    family_chain[
        family_chain["mapping_type"] != "single"
    ]
)
```

```python
# ============================================================
# MODULE 2A — Pull all individual contract metadata
# ============================================================

selected_codes = sorted(
    family_chain["lseg_contrcode"]
    .dropna()
    .astype(int)
    .unique()
)

print("Unique LSEG families:", len(selected_codes))

codes_sql = ",".join(map(str, selected_codes))

contracts_raw = db.raw_sql(f"""
SELECT DISTINCT
    contrcode,
    futcode,
    clscode,
    dsmnem,
    contrdate,
    startdate,
    lasttrddate,
    expirationdate,
    exchtickersymb,
    ldb
FROM tr_ds_fut.wrds_contract_info
WHERE contrcode IN ({codes_sql})
ORDER BY contrcode, contrdate, futcode
""")

print("Metadata rows:", len(contracts_raw))
print("Unique futcodes:", contracts_raw["futcode"].nunique())


# ------------------------------------------------------------
# QA 1 — How many individual contracts per family?
# ------------------------------------------------------------

contract_summary = (
    contracts_raw
    .groupby("contrcode")
    .agg(
        n_futcodes=("futcode", "nunique"),
        first_start=("startdate", "min"),
        last_end=("lasttrddate", "max"),
        missing_ltd=("lasttrddate", lambda x: x.isna().sum()),
    )
    .reset_index()
)

display(contract_summary)


# ------------------------------------------------------------
# QA 2 — Does the same futcode appear multiple times?
# ------------------------------------------------------------

dup_futcodes = (
    contracts_raw
    .groupby(["contrcode", "futcode"])
    .agg(
        n_rows=("futcode", "size"),
        n_ldbs=("ldb", "nunique"),
        ldbs=("ldb", lambda x: sorted(set(x.dropna()))),
    )
    .reset_index()
)

dup_futcodes = dup_futcodes[
    dup_futcodes["n_rows"] > 1
]

print("Duplicated contrcode/futcode pairs:", len(dup_futcodes))

display(dup_futcodes.head(50))


# ------------------------------------------------------------
# QA 3 — Contracts with missing expiry / last trade information
# ------------------------------------------------------------

missing_dates = contracts_raw[
    contracts_raw["lasttrddate"].isna()
    | contracts_raw["contrdate"].isna()
]

print("Contracts missing contrdate or LTD:", len(missing_dates))

display(missing_dates.head(50))
```

```python
# ============================================================
# MODULE 2B — Keep only contracts relevant to paper + OOS period
# ============================================================

DATA_END = pd.Timestamp("2026-10-31")

# Add paper start/end to each family segment
family_periods = family_chain.merge(
    paper[
        [
            "paper_ticker",
            "paper_start",
            "paper_end",
        ]
    ],
    on="paper_ticker",
    how="left"
)

family_periods["desired_start"] = pd.to_datetime(
    family_periods["paper_start"].astype(int).astype(str) + "-01-01"
)

family_periods["desired_end"] = DATA_END


# ------------------------------------------------------------
# Attach every individual contract to the paper market(s)
# that may use that family
# ------------------------------------------------------------

contracts_needed = family_periods.merge(
    contracts_raw,
    left_on="lseg_contrcode",
    right_on="contrcode",
    how="left"
)

contracts_needed["startdate"] = pd.to_datetime(
    contracts_needed["startdate"]
)

contracts_needed["lasttrddate"] = pd.to_datetime(
    contracts_needed["lasttrddate"]
)


# Keep contracts that could overlap the desired period.
# If LTD is missing, keep it for now; actual price history
# will determine its effective last trading date.
contracts_needed = contracts_needed[
    (contracts_needed["startdate"] <= contracts_needed["desired_end"])
    &
    (
        contracts_needed["lasttrddate"].isna()
        |
        (contracts_needed["lasttrddate"] >= contracts_needed["desired_start"])
    )
].copy()


print("Paper markets represented:",
      contracts_needed["paper_ticker"].nunique())

print("Unique families represented:",
      contracts_needed["contrcode"].nunique())

print("Unique futcodes needed:",
      contracts_needed["futcode"].nunique())


# ------------------------------------------------------------
# Missing LTD diagnostics
# ------------------------------------------------------------

missing_ltd_summary = (
    contracts_needed[
        contracts_needed["lasttrddate"].isna()
    ]
    .groupby(
        ["paper_ticker", "contrcode", "exchtickersymb"]
    )
    .agg(
        missing_ltd=("futcode", "nunique"),
        earliest_start=("startdate", "min"),
        latest_start=("startdate", "max"),
    )
    .reset_index()
)

print("\nMissing-LTD contracts after period filter:")
print(
    contracts_needed["lasttrddate"]
    .isna()
    .sum()
)

display(missing_ltd_summary)
```

```python
# ============================================================
# MODULE 2C — Clean missing-LTD contracts and deduplicate futcodes
# ============================================================

def infer_contract_month(row):
    """
    contrdate looks like MMYY, e.g.
    0199 = Jan 1999
    0301 = Mar 2001
    1226 = Dec 2026

    Use startdate to infer 19xx vs 20xx.
    """
    x = str(row["contrdate"]).strip().zfill(4)

    if len(x) != 4 or not x.isdigit():
        return pd.NaT

    month = int(x[:2])
    yy = int(x[2:])

    if month < 1 or month > 12:
        return pd.NaT

    start_year = pd.to_datetime(row["startdate"]).year

    candidates = [1900 + yy, 2000 + yy]

    year = min(
        candidates,
        key=lambda y: abs(y - start_year)
    )

    return pd.Timestamp(year=year, month=month, day=1)


contracts_needed["contract_month"] = contracts_needed.apply(
    infer_contract_month,
    axis=1
)


# ------------------------------------------------------------
# For missing LTD:
# drop contract if its contract month is already before
# the paper market's required start.
#
# If contract_month itself cannot be parsed, keep it for safety.
# ------------------------------------------------------------

keep_mask = (
    contracts_needed["lasttrddate"].notna()
    |
    contracts_needed["contract_month"].isna()
    |
    (
        contracts_needed["contract_month"]
        >= contracts_needed["desired_start"]
    )
)

contracts_clean = contracts_needed[keep_mask].copy()


# ------------------------------------------------------------
# Deduplicate shared families such as:
# CL / ZU and NR / ZN
# ------------------------------------------------------------

download_contracts = (
    contracts_clean
    .groupby(["contrcode", "futcode"], as_index=False)
    .agg(
        dsmnem=("dsmnem", "first"),
        contrdate=("contrdate", "first"),
        contract_month=("contract_month", "first"),
        startdate=("startdate", "first"),
        lasttrddate=("lasttrddate", "first"),
        expirationdate=("expirationdate", "first"),
        exchtickersymb=("exchtickersymb", "first"),
        ldb=("ldb", "first"),
        desired_start=("desired_start", "min"),
        desired_end=("desired_end", "max"),
        paper_markets=(
            "paper_ticker",
            lambda x: ",".join(sorted(set(x)))
        ),
    )
)


print("Unique futcodes before cleaning:",
      contracts_needed["futcode"].nunique())

print("Unique futcodes after cleaning:",
      download_contracts["futcode"].nunique())

print(
    "Remaining unique contracts with missing LTD:",
    download_contracts["lasttrddate"].isna().sum()
)

print(
    "Unparsed contract months:",
    download_contracts["contract_month"].isna().sum()
)


# ------------------------------------------------------------
# Show where missing LTD still remains
# ------------------------------------------------------------

display(
    download_contracts[
        download_contracts["lasttrddate"].isna()
    ]
    .groupby(
        ["contrcode", "exchtickersymb", "paper_markets"]
    )
    .agg(
        n_missing=("futcode", "nunique"),
        first_contract_month=("contract_month", "min"),
        last_contract_month=("contract_month", "max"),
    )
    .reset_index()
)
```

```python
# ============================================================
# MODULE 2D — Fill missing LTD from actual settlement history
# ============================================================

missing_futcodes = (
    download_contracts.loc[
        download_contracts["lasttrddate"].isna(),
        "futcode"
    ]
    .dropna()
    .astype(int)
    .unique()
)

print("Missing-LTD futcodes to resolve:", len(missing_futcodes))

missing_sql = ",".join(map(str, missing_futcodes))

actual_history = db.raw_sql(f"""
SELECT
    futcode,
    MIN(date_) AS first_settlement_date,
    MAX(date_) AS last_settlement_date,
    COUNT(*) AS n_settlement_rows
FROM tr_ds_fut.wrds_fut_contract
WHERE futcode IN ({missing_sql})
  AND settlement IS NOT NULL
GROUP BY futcode
""")

actual_history["first_settlement_date"] = pd.to_datetime(
    actual_history["first_settlement_date"]
)

actual_history["last_settlement_date"] = pd.to_datetime(
    actual_history["last_settlement_date"]
)

print(
    "Missing-LTD contracts with settlement history:",
    actual_history["futcode"].nunique()
)

print(
    "Missing-LTD contracts with NO settlement history:",
    len(missing_futcodes)
    - actual_history["futcode"].nunique()
)


# ------------------------------------------------------------
# Add actual history back to contract metadata
# ------------------------------------------------------------

download_contracts = download_contracts.merge(
    actual_history,
    on="futcode",
    how="left"
)

download_contracts["effective_ltd"] = (
    download_contracts["lasttrddate"]
    .fillna(download_contracts["last_settlement_date"])
)


# ------------------------------------------------------------
# QA
# ------------------------------------------------------------

print(
    "Contracts still missing effective LTD:",
    download_contracts["effective_ltd"].isna().sum()
)

display(
    download_contracts[
        download_contracts["lasttrddate"].isna()
    ][[
        "contrcode",
        "futcode",
        "dsmnem",
        "contract_month",
        "first_settlement_date",
        "last_settlement_date",
        "effective_ltd",
        "paper_markets",
    ]]
    .head(50)
)
```

```python
# ============================================================
# MODULE 2G — Diagnose missing futcodes + duplicate daily rows
# ============================================================

# ------------------------------------------------------------
# 1. Where do the 167 missing futcodes come from?
# ------------------------------------------------------------

missing_meta = download_contracts[
    download_contracts["futcode"]
    .astype(int)
    .isin(missing_futcodes)
].copy()

print("Missing futcodes by market/family:")

display(
    missing_meta
    .groupby(
        ["paper_markets", "contrcode", "exchtickersymb"]
    )
    .agg(
        n_missing=("futcode", "nunique"),
        first_contract=("contract_month", "min"),
        last_contract=("contract_month", "max"),
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Are duplicate futcode/date rows identical or conflicting?
# ------------------------------------------------------------

exact_duplicate_rows = 0
duplicate_pairs = 0
conflicting_pairs = 0

conflict_examples = []

for f in files:

    x = pd.read_parquet(
        f,
        columns=[
            "futcode",
            "date_",
            "settlement",
            "volume",
            "openinterest",
        ]
    )

    dup = x[
        x.duplicated(
            ["futcode", "date_"],
            keep=False
        )
    ]

    if len(dup) == 0:
        continue

    # completely identical rows
    exact_duplicate_rows += dup.duplicated(
        [
            "futcode",
            "date_",
            "settlement",
            "volume",
            "openinterest",
        ],
        keep="first"
    ).sum()

    stats = (
        dup
        .groupby(["futcode", "date_"])
        .agg(
            n_rows=("futcode", "size"),
            n_settlement=("settlement",
                          lambda s: s.nunique(dropna=False)),
            n_volume=("volume",
                      lambda s: s.nunique(dropna=False)),
            n_openinterest=("openinterest",
                            lambda s: s.nunique(dropna=False)),
        )
        .reset_index()
    )

    duplicate_pairs += len(stats)

    conflicts = stats[
        (stats["n_settlement"] > 1)
        | (stats["n_volume"] > 1)
        | (stats["n_openinterest"] > 1)
    ]

    conflicting_pairs += len(conflicts)

    if len(conflicts) > 0 and len(conflict_examples) < 20:
        conflict_examples.append(
            conflicts.head(20)
        )


print("\nDuplicate futcode/date pairs:", duplicate_pairs)
print("Exact duplicate extra rows:", exact_duplicate_rows)
print("Conflicting futcode/date pairs:", conflicting_pairs)

if conflict_examples:
    display(
        pd.concat(conflict_examples)
        .head(30)
    )
```

```python
# ============================================================
# MODULE 3A — Compute LTD-5 trading-day roll date
# ============================================================

import pandas as pd
import pyarrow.parquet as pq

DATA_DIR = Path(
    "/Users/komali/Desktop/Graduate/2026 Fall/"
    "Data Driven Method/Final Project/Data"
)

CLEAN_FILE = DATA_DIR / "futures_daily_clean.parquet"

# only contracts that actually have settlement data
roll_meta = contract_metadata_clean[
    contract_metadata_clean["has_settlement"]
].copy()

roll_meta["effective_ltd"] = pd.to_datetime(
    roll_meta["effective_ltd"]
)

# ------------------------------------------------------------
# Read only columns needed for roll-date construction
# ------------------------------------------------------------

daily = pd.read_parquet(
    CLEAN_FILE,
    columns=[
        "futcode",
        "date_",
        "settlement",
    ]
)

daily["date_"] = pd.to_datetime(daily["date_"])

daily = daily[
    daily["settlement"].notna()
].copy()


# ------------------------------------------------------------
# For each contract:
# find trading dates <= effective LTD,
# then choose 5 trading sessions before LTD anchor
# ------------------------------------------------------------

roll_rows = []

for futcode, g in daily.groupby("futcode"):

    meta = roll_meta[
        roll_meta["futcode"] == futcode
    ]

    if len(meta) == 0:
        continue

    ltd = meta["effective_ltd"].iloc[0]

    dates = (
        g.loc[g["date_"] <= ltd, "date_"]
        .drop_duplicates()
        .sort_values()
        .to_numpy()
    )

    if len(dates) == 0:
        continue

    # actual last observed trading day <= LTD
    ltd_anchor = pd.Timestamp(dates[-1])

    # 5 trading sessions BEFORE the anchor
    if len(dates) >= 6:
        roll_date = pd.Timestamp(dates[-6])
    else:
        roll_date = pd.NaT

    roll_rows.append({
        "futcode": int(futcode),
        "effective_ltd": ltd,
        "ltd_anchor": ltd_anchor,
        "roll_date": roll_date,
        "n_trading_days": len(dates),
    })


roll_schedule = pd.DataFrame(roll_rows)

# attach contract metadata
roll_schedule = (
    roll_schedule
    .merge(
        roll_meta[[
            "futcode",
            "contrcode",
            "contract_month",
            "dsmnem",
            "exchtickersymb",
        ]],
        on="futcode",
        how="left"
    )
)


# ------------------------------------------------------------
# QA
# ------------------------------------------------------------

print("Contracts with settlement:",
      roll_meta["futcode"].nunique())

print("Roll dates constructed:",
      roll_schedule["roll_date"].notna().sum())

print("Missing roll dates:",
      roll_schedule["roll_date"].isna().sum())

print("\nSample:")
display(
    roll_schedule[
        roll_schedule["roll_date"].notna()
    ]
    .sort_values(["contrcode", "contract_month"])
    .head(30)
)
```

```python
# ============================================================
# MODULE 3A.2 — Final robust LTD-5 roll schedule
# ============================================================

roll_meta_final = contract_metadata_clean[
    contract_metadata_clean["has_settlement"]
].copy()

for c in [
    "effective_ltd",
    "actual_first_date",
    "actual_last_date",
]:
    roll_meta_final[c] = pd.to_datetime(
        roll_meta_final[c]
    )


# ------------------------------------------------------------
# Correct obviously invalid metadata LTD:
# LTD earlier than the first actual settlement
# ------------------------------------------------------------

bad_ltd = (
    roll_meta_final["effective_ltd"]
    < roll_meta_final["actual_first_date"]
)

print("Invalid metadata LTD:", bad_ltd.sum())

roll_meta_final["roll_ltd"] = (
    roll_meta_final["effective_ltd"]
)

roll_meta_final.loc[
    bad_ltd,
    "roll_ltd"
] = roll_meta_final.loc[
    bad_ltd,
    "actual_last_date"
]


# ------------------------------------------------------------
# Build roll dates again
# ------------------------------------------------------------

roll_rows = []

for futcode, g in daily.groupby("futcode"):

    meta = roll_meta_final[
        roll_meta_final["futcode"] == futcode
    ]

    if len(meta) == 0:
        continue

    ltd = meta["roll_ltd"].iloc[0]

    dates = (
        g.loc[
            g["date_"] <= ltd,
            "date_"
        ]
        .drop_duplicates()
        .sort_values()
        .to_numpy()
    )

    if len(dates) == 0:
        roll_date = pd.NaT
        ltd_anchor = pd.NaT

    else:
        ltd_anchor = pd.Timestamp(dates[-1])

        if len(dates) >= 6:
            roll_date = pd.Timestamp(dates[-6])
        else:
            roll_date = pd.NaT

    roll_rows.append({
        "futcode": int(futcode),
        "roll_ltd": ltd,
        "ltd_anchor": ltd_anchor,
        "roll_date": roll_date,
        "n_trading_days": len(dates),
    })


roll_schedule_final = pd.DataFrame(
    roll_rows
)

roll_schedule_final = (
    roll_schedule_final
    .merge(
        roll_meta_final[[
            "futcode",
            "contrcode",
            "contract_month",
            "dsmnem",
            "exchtickersymb",
        ]],
        on="futcode",
        how="left"
    )
)

roll_schedule_final["roll_eligible"] = (
    roll_schedule_final["roll_date"].notna()
)


# save
ROLL_FILE = DATA_DIR / "roll_schedule.parquet"

roll_schedule_final.to_parquet(
    ROLL_FILE,
    index=False
)


print("\nContracts:", len(roll_schedule_final))

print(
    "Roll eligible:",
    roll_schedule_final["roll_eligible"].sum()
)

print(
    "Excluded thin contracts:",
    (~roll_schedule_final["roll_eligible"]).sum()
)

print("\nSaved:")
print(ROLL_FILE)
```

```python
# ============================================================
# MODULE 3B — Build contract roll transitions + overlap QA
# ============================================================

# only contracts with valid LTD-5 roll dates
eligible = (
    roll_schedule_final[
        roll_schedule_final["roll_eligible"]
    ]
    .copy()
    .sort_values(
        ["contrcode", "contract_month", "roll_ltd"]
    )
)

# ------------------------------------------------------------
# Next eligible contract within each LSEG family
# ------------------------------------------------------------

eligible["next_futcode"] = (
    eligible
    .groupby("contrcode")["futcode"]
    .shift(-1)
)

eligible["next_contract_month"] = (
    eligible
    .groupby("contrcode")["contract_month"]
    .shift(-1)
)

# every contract except the final one in each family
transitions = eligible[
    eligible["next_futcode"].notna()
].copy()

transitions["next_futcode"] = (
    transitions["next_futcode"]
    .astype("int64")
)

# ------------------------------------------------------------
# Settlement lookup
# ------------------------------------------------------------

price_lookup = daily[[
    "futcode",
    "date_",
    "settlement"
]].copy()

price_lookup["futcode"] = (
    price_lookup["futcode"].astype("int64")
)

# old contract settlement on roll date
old_px = price_lookup.rename(
    columns={
        "futcode": "futcode",
        "date_": "roll_date",
        "settlement": "old_settlement",
    }
)

transitions = transitions.merge(
    old_px,
    on=["futcode", "roll_date"],
    how="left"
)

# new contract settlement on the SAME roll date
new_px = price_lookup.rename(
    columns={
        "futcode": "next_futcode",
        "date_": "roll_date",
        "settlement": "new_settlement",
    }
)

transitions = transitions.merge(
    new_px,
    on=["next_futcode", "roll_date"],
    how="left"
)

# ------------------------------------------------------------
# QA
# ------------------------------------------------------------

transitions["same_day_overlap"] = (
    transitions["old_settlement"].notna()
    & transitions["new_settlement"].notna()
)

print("Total roll transitions:",
      len(transitions))

print(
    "Same-day old/new settlement:",
    transitions["same_day_overlap"].sum()
)

print(
    "Missing same-day overlap:",
    (~transitions["same_day_overlap"]).sum()
)

display(
    transitions.loc[
        ~transitions["same_day_overlap"],
        [
            "contrcode",
            "futcode",
            "contract_month",
            "next_futcode",
            "next_contract_month",
            "roll_date",
            "old_settlement",
            "new_settlement",
        ]
    ].head(50)
)
```

```python
# ============================================================
# MODULE 3B.1 — Final robust roll transitions
# ============================================================

# ------------------------------------------------------------
# 1. Order contracts by actual roll chronology
# ------------------------------------------------------------

eligible = (
    roll_schedule_final[
        roll_schedule_final["roll_eligible"]
    ]
    .copy()
    .sort_values([
        "contrcode",
        "roll_date",
        "roll_ltd",
        "contract_month",
    ])
)

# ------------------------------------------------------------
# 2. Build settlement lookup
# ------------------------------------------------------------

price_data = daily[
    daily["settlement"].notna()
][[
    "futcode",
    "date_",
    "settlement",
]].copy()

price_data["futcode"] = (
    price_data["futcode"].astype("int64")
)

price_data["date_"] = pd.to_datetime(
    price_data["date_"]
)

price_lookup = {
    int(fc): dict(zip(g["date_"], g["settlement"]))
    for fc, g in price_data.groupby("futcode")
}


# ------------------------------------------------------------
# 3. At old contract's T-5:
# choose nearest later contract with settlement that day
# ------------------------------------------------------------

transition_rows = []

for contrcode, g in eligible.groupby("contrcode"):

    g = g.reset_index(drop=True)

    for i in range(len(g) - 1):

        old = g.iloc[i]

        old_fc = int(old["futcode"])
        roll_date = pd.Timestamp(old["roll_date"])

        old_px = (
            price_lookup
            .get(old_fc, {})
            .get(roll_date)
        )

        chosen = None
        steps_ahead = None

        # Search subsequent contracts in actual roll order
        for j in range(i + 1, len(g)):

            candidate = g.iloc[j]
            new_fc = int(candidate["futcode"])

            new_px = (
                price_lookup
                .get(new_fc, {})
                .get(roll_date)
            )

            if new_px is not None:
                chosen = candidate
                steps_ahead = j - i
                break

        if chosen is None:

            transition_rows.append({
                "contrcode": contrcode,
                "old_futcode": old_fc,
                "old_contract_month": old["contract_month"],
                "roll_date": roll_date,

                "new_futcode": pd.NA,
                "new_contract_month": pd.NaT,

                "old_settlement": old_px,
                "new_settlement": pd.NA,

                "steps_ahead": pd.NA,
                "same_day_overlap": False,
                "roll_ratio": pd.NA,
            })

        else:

            new_fc = int(chosen["futcode"])

            new_px = (
                price_lookup[new_fc][roll_date]
            )

            transition_rows.append({
                "contrcode": contrcode,
                "old_futcode": old_fc,
                "old_contract_month": old["contract_month"],
                "roll_date": roll_date,

                "new_futcode": new_fc,
                "new_contract_month": chosen["contract_month"],

                "old_settlement": old_px,
                "new_settlement": new_px,

                "steps_ahead": steps_ahead,
                "same_day_overlap": True,

                "roll_ratio":
                    new_px / old_px
                    if old_px not in [None, 0]
                    else pd.NA,
            })


transitions_final = pd.DataFrame(
    transition_rows
)


# ------------------------------------------------------------
# 4. Save
# ------------------------------------------------------------

TRANSITION_FILE = (
    DATA_DIR / "roll_transitions.parquet"
)

transitions_final.to_parquet(
    TRANSITION_FILE,
    index=False
)


# ------------------------------------------------------------
# 5. QA
# ------------------------------------------------------------

print(
    "Total transitions:",
    len(transitions_final)
)

print(
    "Valid same-day transitions:",
    transitions_final["same_day_overlap"].sum()
)

print(
    "Unresolved transitions:",
    (~transitions_final["same_day_overlap"]).sum()
)

print("\nSkipped-expiry distribution:")

print(
    transitions_final["steps_ahead"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nUnresolved:")

display(
    transitions_final[
        ~transitions_final["same_day_overlap"]
    ]
)
```

```python
# ============================================================
# MODULE 3A.3 — Robust LTD + FAMILY trading-calendar T-5
# ============================================================

import numpy as np
import pandas as pd

meta = contract_metadata_clean[
    contract_metadata_clean["has_settlement"]
].copy()

meta["futcode"] = meta["futcode"].astype("int64")

for c in [
    "contract_month",
    "lasttrddate",
    "actual_first_date",
    "actual_last_date",
]:
    meta[c] = pd.to_datetime(
        meta[c],
        errors="coerce"
    )


# ============================================================
# 1. Settlement data + family membership
# ============================================================

daily_valid = (
    daily[
        daily["settlement"].notna()
    ][[
        "futcode",
        "date_",
        "settlement",
    ]]
    .copy()
)

daily_valid["futcode"] = (
    daily_valid["futcode"].astype("int64")
)

daily_valid["date_"] = pd.to_datetime(
    daily_valid["date_"]
)

fc_family = (
    meta[[
        "futcode",
        "contrcode",
    ]]
    .drop_duplicates()
)

daily_family = daily_valid.merge(
    fc_family,
    on="futcode",
    how="inner"
)


# ============================================================
# 2. Build ROBUST LTD
# ============================================================

# A normal LTD should be reasonably close to contract month.
# +/-180 days is deliberately conservative.
meta["metadata_ltd_gap"] = (
    meta["lasttrddate"]
    - meta["contract_month"]
).dt.days

meta["metadata_ltd_valid"] = (
    meta["lasttrddate"].notna()
    &
    meta["metadata_ltd_gap"].abs().le(180)
)

meta["robust_ltd"] = pd.NaT

# use trustworthy metadata LTD
meta.loc[
    meta["metadata_ltd_valid"],
    "robust_ltd"
] = meta.loc[
    meta["metadata_ltd_valid"],
    "lasttrddate"
]


# ------------------------------------------------------------
# Fallback:
# if LTD missing / obviously wrong,
# use latest actual settlement within +/-90 days
# of the stated contract month
# ------------------------------------------------------------

dates_by_fc = {
    int(fc): pd.DatetimeIndex(
        g["date_"].sort_values().unique()
    )
    for fc, g in daily_valid.groupby("futcode")
}

fallback_count = 0

for idx, row in meta[
    ~meta["metadata_ltd_valid"]
].iterrows():

    fc = int(row["futcode"])
    cm = row["contract_month"]

    if pd.isna(cm):
        continue

    dates = dates_by_fc.get(
        fc,
        pd.DatetimeIndex([])
    )

    near = dates[
        (dates >= cm - pd.Timedelta(days=90))
        &
        (dates <= cm + pd.Timedelta(days=90))
    ]

    if len(near) > 0:

        meta.at[
            idx,
            "robust_ltd"
        ] = near.max()

        fallback_count += 1


# ============================================================
# 3. Last actual settlement BEFORE robust LTD
# ============================================================

anchors = []

for _, row in meta.iterrows():

    fc = int(row["futcode"])
    ltd = row["robust_ltd"]

    if pd.isna(ltd):
        anchors.append(pd.NaT)
        continue

    dates = dates_by_fc.get(
        fc,
        pd.DatetimeIndex([])
    )

    valid_dates = dates[
        dates <= ltd
    ]

    if len(valid_dates) == 0:
        anchors.append(pd.NaT)
    else:
        anchors.append(valid_dates.max())


meta["ltd_anchor"] = anchors

meta["anchor_gap_days"] = (
    meta["robust_ltd"]
    - meta["ltd_anchor"]
).dt.days


# ============================================================
# 4. FAMILY trading calendar
# ============================================================

family_calendars = {
    cc: pd.DatetimeIndex(
        g["date_"]
        .drop_duplicates()
        .sort_values()
    )
    for cc, g in daily_family.groupby("contrcode")
}


# ============================================================
# 5. T-5 on FAMILY calendar
# ============================================================

roll_dates = []

for _, row in meta.iterrows():

    cc = row["contrcode"]
    ltd = row["robust_ltd"]

    if pd.isna(ltd):
        roll_dates.append(pd.NaT)
        continue

    calendar = family_calendars.get(
        cc,
        pd.DatetimeIndex([])
    )

    dates = calendar[
        calendar <= ltd
    ]

    if len(dates) >= 6:
        roll_dates.append(
            dates[-6]
        )
    else:
        roll_dates.append(
            pd.NaT
        )


meta["roll_date"] = roll_dates


# ============================================================
# 6. Require the contract to still be alive near LTD
# ============================================================

# 14 calendar days is conservative for T-5 trading sessions.
meta["near_ltd"] = (
    meta["anchor_gap_days"]
    .between(0, 14)
)

meta["roll_eligible_v2"] = (
    meta["robust_ltd"].notna()
    &
    meta["roll_date"].notna()
    &
    meta["near_ltd"]
)


# ============================================================
# 7. Does OLD contract itself have settlement on exact T-5?
# ============================================================

def has_price(fc, dt):

    if pd.isna(dt):
        return False

    dates = dates_by_fc.get(
        int(fc),
        pd.DatetimeIndex([])
    )

    return dt in dates


meta["old_has_roll_price"] = [
    has_price(fc, dt)
    for fc, dt in zip(
        meta["futcode"],
        meta["roll_date"]
    )
]


# ============================================================
# 8. Save V2
# ============================================================

ROLL_V2_FILE = (
    DATA_DIR / "roll_schedule_v2.parquet"
)

meta.to_parquet(
    ROLL_V2_FILE,
    index=False
)


# ============================================================
# QA
# ============================================================

print(
    "Contracts:",
    len(meta)
)

print(
    "Trusted metadata LTD:",
    meta["metadata_ltd_valid"].sum()
)

print(
    "Fallback LTD:",
    fallback_count
)

print(
    "No plausible LTD:",
    meta["robust_ltd"].isna().sum()
)

print(
    "Near-LTD settlement:",
    meta["near_ltd"].sum()
)

print(
    "Final roll eligible:",
    meta["roll_eligible_v2"].sum()
)

print(
    "Eligible with OLD price on exact T-5:",
    (
        meta["roll_eligible_v2"]
        &
        meta["old_has_roll_price"]
    ).sum()
)

print(
    "Eligible but OLD price missing on T-5:",
    (
        meta["roll_eligible_v2"]
        &
        ~meta["old_has_roll_price"]
    ).sum()
)

print("\nSaved:")
print(ROLL_V2_FILE)
```

```python
# ============================================================
# MODULE 3A.4 — Final roll schedule with right-censoring
# ============================================================

roll_v3 = meta.copy()

DATA_END = daily_valid["date_"].max()

print("Data end:", DATA_END)

# ------------------------------------------------------------
# 1. Matured vs right-censored
# ------------------------------------------------------------

roll_v3["is_matured"] = (
    roll_v3["robust_ltd"].notna()
    & (roll_v3["robust_ltd"] <= DATA_END)
)

roll_v3["is_right_censored"] = (
    roll_v3["robust_ltd"].notna()
    & (roll_v3["robust_ltd"] > DATA_END)
)


# ------------------------------------------------------------
# 2. Only matured contracts get a T-5 roll date
# ------------------------------------------------------------

roll_v3["roll_date_v3"] = pd.NaT

for idx, row in roll_v3[
    roll_v3["is_matured"]
].iterrows():

    cc = row["contrcode"]
    ltd = row["robust_ltd"]

    calendar = family_calendars.get(
        cc,
        pd.DatetimeIndex([])
    )

    dates = calendar[
        calendar <= ltd
    ]

    if len(dates) >= 6:
        roll_v3.at[
            idx,
            "roll_date_v3"
        ] = dates[-6]


# ------------------------------------------------------------
# 3. Does OLD contract have a settlement on exact T-5?
# ------------------------------------------------------------

roll_v3["old_has_roll_price_v3"] = [
    has_price(fc, dt)
    for fc, dt in zip(
        roll_v3["futcode"],
        roll_v3["roll_date_v3"]
    )
]


# ------------------------------------------------------------
# 4. Valid contract versus executable roll
# ------------------------------------------------------------

# Historical contract is trustworthy only if it remained alive near LTD.
# Future/right-censored contracts are allowed as possible next contracts.
roll_v3["contract_valid_v3"] = (
    roll_v3["robust_ltd"].notna()
    &
    (
        roll_v3["near_ltd"]
        |
        roll_v3["is_right_censored"]
    )
)

# A historical roll is scheduled at exact family-calendar T-5
roll_v3["roll_scheduled_v3"] = (
    roll_v3["is_matured"]
    &
    roll_v3["near_ltd"]
    &
    roll_v3["roll_date_v3"].notna()
)

# Executable means OLD settlement actually exists on exact T-5
roll_v3["roll_executable_v3"] = (
    roll_v3["roll_scheduled_v3"]
    &
    roll_v3["old_has_roll_price_v3"]
)


# ------------------------------------------------------------
# 5. Save
# ------------------------------------------------------------

ROLL_V3_FILE = (
    DATA_DIR / "roll_schedule_v3.parquet"
)

roll_v3.to_parquet(
    ROLL_V3_FILE,
    index=False
)


# ------------------------------------------------------------
# QA
# ------------------------------------------------------------

print("\nContracts:", len(roll_v3))

print(
    "Matured contracts:",
    roll_v3["is_matured"].sum()
)

print(
    "Right-censored contracts:",
    roll_v3["is_right_censored"].sum()
)

print(
    "No plausible LTD:",
    roll_v3["robust_ltd"].isna().sum()
)

print(
    "Valid contracts:",
    roll_v3["contract_valid_v3"].sum()
)

print(
    "Scheduled historical T-5 rolls:",
    roll_v3["roll_scheduled_v3"].sum()
)

print(
    "Executable exact T-5 rolls:",
    roll_v3["roll_executable_v3"].sum()
)

print(
    "Scheduled but OLD price missing:",
    (
        roll_v3["roll_scheduled_v3"]
        & ~roll_v3["old_has_roll_price_v3"]
    ).sum()
)

print(
    "Future contracts incorrectly assigned roll date:",
    (
        roll_v3["is_right_censored"]
        & roll_v3["roll_date_v3"].notna()
    ).sum()
)

print("\nSaved:")
print(ROLL_V3_FILE)
```

```python
# ============================================================
# MODULE 3B.2 — Final V3 roll transitions
# ============================================================

# ------------------------------------------------------------
# 1. Contracts allowed in the roll chain
# ------------------------------------------------------------

roll_v3["chain_eligible"] = (
    roll_v3["roll_executable_v3"]
    |
    roll_v3["is_right_censored"]
)

chain_contracts = (
    roll_v3[
        roll_v3["chain_eligible"]
    ]
    .copy()
    .sort_values([
        "contrcode",
        "robust_ltd",
        "contract_month",
        "futcode",
    ])
)


# ------------------------------------------------------------
# 2. Price lookup
# ------------------------------------------------------------

price_lookup_v3 = {
    int(fc): dict(
        zip(
            g["date_"],
            g["settlement"]
        )
    )
    for fc, g in daily_valid.groupby("futcode")
}


# ------------------------------------------------------------
# 3. Exact T-5 transitions
# ------------------------------------------------------------

transition_rows = []

for contrcode, g in chain_contracts.groupby("contrcode"):

    g = (
        g.sort_values([
            "robust_ltd",
            "contract_month",
            "futcode",
        ])
        .reset_index(drop=True)
    )

    # only matured contracts actually generate a roll event
    for i in range(len(g)):

        old = g.iloc[i]

        if not old["roll_executable_v3"]:
            continue

        old_fc = int(old["futcode"])
        roll_date = pd.Timestamp(
            old["roll_date_v3"]
        )

        old_px = (
            price_lookup_v3
            .get(old_fc, {})
            .get(roll_date)
        )

        chosen = None
        chosen_px = None
        steps_ahead = None

        # nearest later-expiring valid contract
        for j in range(i + 1, len(g)):

            candidate = g.iloc[j]

            # strictly later expiry
            if (
                candidate["robust_ltd"]
                <= old["robust_ltd"]
            ):
                continue

            new_fc = int(
                candidate["futcode"]
            )

            new_px = (
                price_lookup_v3
                .get(new_fc, {})
                .get(roll_date)
            )

            if new_px is not None:
                chosen = candidate
                chosen_px = new_px
                steps_ahead = j - i
                break

        if chosen is None:

            transition_rows.append({
                "contrcode": contrcode,

                "old_futcode": old_fc,
                "old_contract_month":
                    old["contract_month"],
                "old_ltd":
                    old["robust_ltd"],

                "roll_date": roll_date,

                "new_futcode": pd.NA,
                "new_contract_month": pd.NaT,
                "new_ltd": pd.NaT,

                "old_settlement": old_px,
                "new_settlement": pd.NA,

                "steps_ahead": pd.NA,
                "resolved": False,
                "roll_ratio": pd.NA,
            })

        else:

            new_fc = int(
                chosen["futcode"]
            )

            transition_rows.append({
                "contrcode": contrcode,

                "old_futcode": old_fc,
                "old_contract_month":
                    old["contract_month"],
                "old_ltd":
                    old["robust_ltd"],

                "roll_date": roll_date,

                "new_futcode": new_fc,
                "new_contract_month":
                    chosen["contract_month"],
                "new_ltd":
                    chosen["robust_ltd"],

                "old_settlement": old_px,
                "new_settlement": chosen_px,

                "steps_ahead": steps_ahead,
                "resolved": True,

                "roll_ratio":
                    chosen_px / old_px,
            })


transitions_v3 = pd.DataFrame(
    transition_rows
)


# ------------------------------------------------------------
# 4. Save
# ------------------------------------------------------------

TRANSITIONS_V3_FILE = (
    DATA_DIR / "roll_transitions_v3.parquet"
)

transitions_v3.to_parquet(
    TRANSITIONS_V3_FILE,
    index=False
)


# ------------------------------------------------------------
# 5. QA
# ------------------------------------------------------------

print(
    "Historical executable rolls:",
    roll_v3["roll_executable_v3"].sum()
)

print(
    "Transitions constructed:",
    len(transitions_v3)
)

print(
    "Resolved exact T-5 transitions:",
    transitions_v3["resolved"].sum()
)

print(
    "Unresolved exact T-5 transitions:",
    (~transitions_v3["resolved"]).sum()
)

print("\nSteps-ahead distribution:")

print(
    transitions_v3["steps_ahead"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nUnresolved examples:")

display(
    transitions_v3[
        ~transitions_v3["resolved"]
    ][[
        "contrcode",
        "old_futcode",
        "old_contract_month",
        "old_ltd",
        "roll_date",
    ]]
)
```

```python
# ============================================================
# MODULE 4A — Final family continuous futures
# ============================================================

# only resolved, economically usable roll transitions
good_transitions = transitions_v3[
    transitions_v3["resolved"]
    & (transitions_v3["old_settlement"] > 0)
    & (transitions_v3["new_settlement"] > 0)
    & (transitions_v3["roll_ratio"] > 0)
].copy()

transition_map = {
    int(r["old_futcode"]): r
    for _, r in good_transitions.iterrows()
}

# contracts allowed in the final chain
pool = roll_v3[
    roll_v3["chain_eligible"]
].copy()

price_by_fc = {
    int(fc): g[
        ["date_", "settlement"]
    ].sort_values("date_").copy()
    for fc, g in daily_valid.groupby("futcode")
}

segment_rows = []


for cc, fam in pool.groupby("contrcode"):

    fam = fam.sort_values(
        ["robust_ltd", "contract_month", "futcode"]
    ).copy()

    used = set()
    block_id = 0

    # --------------------------------------------------------
    # Start from earliest usable contract
    # --------------------------------------------------------

    remaining = fam.copy()

    while len(remaining) > 0:

        current = remaining.iloc[0]
        current_fc = int(current["futcode"])

        if current_fc in used:
            remaining = remaining.iloc[1:]
            continue

        block_id += 1
        segment_order = 0
        segment_start = None

        while True:

            if current_fc in used:
                break

            used.add(current_fc)
            segment_order += 1

            px = price_by_fc.get(current_fc)

            if px is None or len(px) == 0:
                break

            if segment_start is None:
                segment_start = px["date_"].min()

            tr = transition_map.get(current_fc)

            # -----------------------------------------------
            # Normal resolved roll
            # -----------------------------------------------

            if tr is not None:

                roll_date = pd.Timestamp(
                    tr["roll_date"]
                )

                segment_rows.append({
                    "contrcode": cc,
                    "block_id": block_id,
                    "segment_order": segment_order,
                    "active_futcode": current_fc,
                    "start_date": segment_start,
                    "end_date": roll_date,
                    "roll_ratio_to_next":
                        float(tr["roll_ratio"]),
                })

                current_fc = int(
                    tr["new_futcode"]
                )

                segment_start = roll_date

                new_row = fam[
                    fam["futcode"].astype(int)
                    == current_fc
                ]

                if len(new_row) == 0:
                    break

                current = new_row.iloc[0]
                continue


            # -----------------------------------------------
            # No valid transition:
            # finish this block — do NOT fabricate a roll
            # -----------------------------------------------

            segment_rows.append({
                "contrcode": cc,
                "block_id": block_id,
                "segment_order": segment_order,
                "active_futcode": current_fc,
                "start_date": segment_start,
                "end_date": pd.NaT,
                "roll_ratio_to_next": pd.NA,
            })

            break


        # ----------------------------------------------------
        # Restart after a gap using next unused later contract
        # ----------------------------------------------------

        used_ltd = fam[
            fam["futcode"].astype(int).isin(used)
        ]["robust_ltd"]

        if len(used_ltd) == 0:
            break

        cutoff = used_ltd.max()

        remaining = fam[
            (~fam["futcode"].astype(int).isin(used))
            & (fam["robust_ltd"] > cutoff)
        ].copy()


segments = pd.DataFrame(segment_rows)


# ============================================================
# Back-adjustment factors by block
# ============================================================

segments["adjustment_factor"] = 1.0

for (cc, block), idx in segments.groupby(
    ["contrcode", "block_id"]
).groups.items():

    ids = list(idx)

    ids = sorted(
        ids,
        key=lambda i: segments.loc[i, "segment_order"]
    )

    running = 1.0

    for i in reversed(ids):

        segments.loc[
            i,
            "adjustment_factor"
        ] = running

        q = segments.loc[
            i,
            "roll_ratio_to_next"
        ]

        if pd.notna(q):
            running *= float(q)


# ============================================================
# Build daily continuous series
# ============================================================

continuous_parts = []

for _, s in segments.iterrows():

    fc = int(s["active_futcode"])

    x = price_by_fc[fc].copy()

    x = x[
        x["date_"] >= s["start_date"]
    ]

    if pd.notna(s["end_date"]):
        x = x[
            x["date_"] < s["end_date"]
        ]

    x["contrcode"] = s["contrcode"]
    x["block_id"] = s["block_id"]
    x["active_futcode"] = fc

    x["adjustment_factor"] = (
        s["adjustment_factor"]
    )

    x["continuous_price"] = (
        x["settlement"]
        * x["adjustment_factor"]
    )

    continuous_parts.append(x)


family_continuous = (
    pd.concat(
        continuous_parts,
        ignore_index=True
    )
    .sort_values(
        ["contrcode", "date_", "block_id"]
    )
)

# one family/date only
family_continuous = (
    family_continuous
    .drop_duplicates(
        ["contrcode", "date_"],
        keep="first"
    )
)


# ============================================================
# Save
# ============================================================

segments.to_parquet(
    DATA_DIR / "family_roll_segments_v3.parquet",
    index=False
)

family_continuous.to_parquet(
    DATA_DIR / "family_continuous_v3.parquet",
    index=False
)


# ============================================================
# QA — only material checks
# ============================================================

print(
    "Families:",
    family_continuous["contrcode"].nunique()
)

print(
    "Continuous observations:",
    f"{len(family_continuous):,}"
)

print(
    "Duplicate family/date:",
    family_continuous.duplicated(
        ["contrcode", "date_"]
    ).sum()
)

print(
    "Missing continuous price:",
    family_continuous[
        "continuous_price"
    ].isna().sum()
)

print(
    "Blocks:",
    segments[
        ["contrcode", "block_id"]
    ].drop_duplicates().shape[0]
)

print(
    "Skipped unresolved rolls:",
    (~transitions_v3["resolved"]).sum()
)

print("\nSaved family_continuous_v3.parquet")
```

```python
# ============================================================
# MODULE 4A.1 — Fix multiplicative back-adjustment
# ============================================================

segments_fixed = segments.copy()
segments_fixed["adjustment_factor"] = 1.0

for (cc, block), idx in segments_fixed.groupby(
    ["contrcode", "block_id"]
).groups.items():

    ids = sorted(
        list(idx),
        key=lambda i:
            segments_fixed.loc[i, "segment_order"]
    )

    running = 1.0

    # newest -> oldest
    for i in reversed(ids):

        q = segments_fixed.loc[
            i,
            "roll_ratio_to_next"
        ]

        # OLD segment must include its OWN roll ratio
        if pd.notna(q):
            running *= float(q)

        segments_fixed.loc[
            i,
            "adjustment_factor"
        ] = running


# ============================================================
# Rebuild continuous prices
# ============================================================

continuous_parts = []

for _, s in segments_fixed.iterrows():

    fc = int(s["active_futcode"])

    x = price_by_fc[fc].copy()

    x = x[
        x["date_"] >= s["start_date"]
    ]

    if pd.notna(s["end_date"]):
        x = x[
            x["date_"] < s["end_date"]
        ]

    x["contrcode"] = s["contrcode"]
    x["block_id"] = s["block_id"]
    x["active_futcode"] = fc
    x["adjustment_factor"] = s["adjustment_factor"]

    x["continuous_price"] = (
        x["settlement"]
        * x["adjustment_factor"]
    )

    continuous_parts.append(x)


family_continuous = (
    pd.concat(
        continuous_parts,
        ignore_index=True
    )
    .sort_values([
        "contrcode",
        "date_",
        "block_id",
    ])
    .drop_duplicates(
        ["contrcode", "date_"],
        keep="first"
    )
)


# ============================================================
# Replace final files
# ============================================================

segments_fixed.to_parquet(
    DATA_DIR / "family_roll_segments_v3.parquet",
    index=False
)

family_continuous.to_parquet(
    DATA_DIR / "family_continuous_v3.parquet",
    index=False
)


print(
    "Families:",
    family_continuous["contrcode"].nunique()
)

print(
    "Observations:",
    f"{len(family_continuous):,}"
)

print(
    "Duplicate family/date:",
    family_continuous.duplicated(
        ["contrcode", "date_"]
    ).sum()
)

print(
    "Missing price:",
    family_continuous["continuous_price"]
    .isna()
    .sum()
)

print("Back-adjustment fixed.")
```

```python
# ============================================================
# MODULE 4B — Build final 64 paper-market continuous series
# ============================================================

import numpy as np
import pandas as pd

fc = family_continuous.copy()

fc["date_"] = pd.to_datetime(fc["date_"])

# actual data end
FINAL_DATA_END = fc["date_"].max()


# ============================================================
# 1. Paper metadata lookup
# ============================================================

paper_meta = (
    paper[[
        "paper_ticker",
        "asset_class",
        "paper_description",
        "paper_start",
        "paper_end",
    ]]
    .drop_duplicates("paper_ticker")
    .set_index("paper_ticker")
)


# ============================================================
# 2. Build each paper market
# ============================================================

market_parts = []
stitch_log = []

for ticker, chain in family_chain.groupby("paper_ticker"):

    chain = chain.sort_values(
        "segment_order"
    ).copy()

    paper_start = pd.Timestamp(
        f"{int(paper_meta.loc[ticker, 'paper_start'])}-01-01"
    )

    ticker_parts = []

    previous_end = None
    stitch_block = 1

    for _, seg in chain.iterrows():

        cc = seg["lseg_contrcode"]

        full_family = (
            fc[
                fc["contrcode"] == cc
            ]
            .copy()
            .sort_values("date_")
        )

        if len(full_family) == 0:
            continue

        # paper start onward
        full_family = full_family[
            full_family["date_"] >= paper_start
        ].copy()

        if len(full_family) == 0:
            continue


        # ----------------------------------------------------
        # predecessor remains active through its last date;
        # successor starts AFTER predecessor's last date
        # ----------------------------------------------------

        part = full_family.copy()

        if previous_end is not None:
            part = part[
                part["date_"] > previous_end
            ].copy()

        if len(part) == 0:
            continue


        # ----------------------------------------------------
        # Cross-family alignment at stitch boundary
        # ----------------------------------------------------

        stitch_ratio = 1.0
        connected = False

        if previous_end is not None and len(ticker_parts) > 0:

            # successor price on predecessor's final date
            new_boundary = full_family[
                full_family["date_"] == previous_end
            ]

            prev_part = ticker_parts[-1]

            old_boundary = prev_part[
                prev_part["date_"] == previous_end
            ]

            if (
                len(new_boundary) > 0
                and len(old_boundary) > 0
            ):

                old_px = float(
                    old_boundary["continuous_price"].iloc[-1]
                )

                new_px = float(
                    new_boundary["continuous_price"].iloc[-1]
                )

                if (
                    np.isfinite(old_px)
                    and np.isfinite(new_px)
                    and old_px > 0
                    and new_px > 0
                ):

                    stitch_ratio = new_px / old_px
                    connected = True

                    # scale only the immediately connected
                    # previous stitch block
                    for k in range(len(ticker_parts)):

                        if (
                            ticker_parts[k]["stitch_block"]
                            .iloc[0]
                            == stitch_block
                        ):
                            ticker_parts[k]["continuous_price"] *= (
                                stitch_ratio
                            )

        # no overlap = new disconnected block
        if (
            previous_end is not None
            and not connected
        ):
            stitch_block += 1


        part["paper_ticker"] = ticker
        part["source_contrcode"] = cc
        part["segment_order"] = seg["segment_order"]
        part["segment_role"] = seg["segment_role"]
        part["mapping_type"] = seg["mapping_type"]
        part["stitch_block"] = stitch_block

        ticker_parts.append(part)

        stitch_log.append({
            "paper_ticker": ticker,
            "segment_order": seg["segment_order"],
            "source_contrcode": cc,
            "segment_role": seg["segment_role"],
            "first_date": part["date_"].min(),
            "last_date": part["date_"].max(),
            "stitch_ratio": stitch_ratio,
            "connected_to_previous": connected,
            "stitch_block": stitch_block,
        })

        # predecessor-until-last-data rule
        previous_end = part["date_"].max()


    if len(ticker_parts) == 0:
        continue

    ticker_df = pd.concat(
        ticker_parts,
        ignore_index=True
    )

    ticker_df = (
        ticker_df
        .sort_values("date_")
        .drop_duplicates(
            ["paper_ticker", "date_"],
            keep="first"
        )
    )

    market_parts.append(ticker_df)


# ============================================================
# 3. Final long-form dataset
# ============================================================

paper64 = (
    pd.concat(
        market_parts,
        ignore_index=True
    )
    .sort_values([
        "paper_ticker",
        "date_",
    ])
    .reset_index(drop=True)
)

stitch_log = pd.DataFrame(stitch_log)


# ============================================================
# 4. Attach paper metadata
# ============================================================

paper64 = paper64.merge(
    paper[[
        "paper_ticker",
        "asset_class",
        "paper_description",
        "paper_start",
        "paper_end",
    ]].drop_duplicates("paper_ticker"),
    on="paper_ticker",
    how="left"
)


# ============================================================
# 5. Wide price matrix
# ============================================================

paper64_prices = (
    paper64
    .pivot(
        index="date_",
        columns="paper_ticker",
        values="continuous_price",
    )
    .sort_index()
)


# ============================================================
# 6. Coverage summary
# ============================================================

coverage = (
    paper64
    .groupby("paper_ticker")
    .agg(
        first_date=("date_", "min"),
        last_date=("date_", "max"),
        observations=("date_", "size"),
        n_stitch_blocks=("stitch_block", "nunique"),
    )
    .reset_index()
    .merge(
        paper[[
            "paper_ticker",
            "asset_class",
            "paper_start",
            "paper_end",
        ]].drop_duplicates("paper_ticker"),
        on="paper_ticker",
        how="left"
    )
)

coverage["desired_start"] = pd.to_datetime(
    coverage["paper_start"]
    .astype(int)
    .astype(str)
    + "-01-01"
)

coverage["start_delay_days"] = (
    coverage["first_date"]
    - coverage["desired_start"]
).dt.days


# ============================================================
# 7. Save EVERYTHING under Data
# ============================================================

paper64.to_parquet(
    DATA_DIR / "paper64_continuous_v3.parquet",
    index=False
)

paper64_prices.to_parquet(
    DATA_DIR / "paper64_prices_v3.parquet"
)

coverage.to_csv(
    DATA_DIR / "paper64_coverage_v3.csv",
    index=False
)

stitch_log.to_csv(
    DATA_DIR / "paper64_stitch_log_v3.csv",
    index=False
)

family_chain.to_csv(
    DATA_DIR / "paper64_family_mapping_v3.csv",
    index=False
)


# ============================================================
# 8. MATERIAL QA ONLY
# ============================================================

print(
    "Paper markets:",
    paper64["paper_ticker"].nunique()
)

print(
    "Rows:",
    f"{len(paper64):,}"
)

print(
    "Duplicate ticker/date:",
    paper64.duplicated(
        ["paper_ticker", "date_"]
    ).sum()
)

print(
    "Missing continuous price:",
    paper64["continuous_price"]
    .isna()
    .sum()
)

print(
    "Wide matrix markets:",
    paper64_prices.shape[1]
)

print(
    "Overall date range:",
    paper64["date_"].min(),
    "to",
    paper64["date_"].max()
)

print("\nLargest start-date shortfalls:")

display(
    coverage
    .sort_values(
        "start_delay_days",
        ascending=False
    )
    .head(15)
)

print("\nSaved final 64-market data under Data/")
```

```python
# ============================================================
# MODULE 4C — Final block-aware daily returns
# ============================================================

import numpy as np
import pandas as pd

final_data = (
    paper64
    .sort_values(["paper_ticker", "date_"])
    .copy()
)

g = final_data.groupby("paper_ticker")

prev_source = g["source_contrcode"].shift(1)
prev_family_block = g["block_id"].shift(1)
prev_stitch_block = g["stitch_block"].shift(1)

# ------------------------------------------------------------
# A new return block starts when:
# 1. first observation of a market
# 2. disconnected cross-family stitch
# 3. unresolved break inside same LSEG family
# ------------------------------------------------------------

first_obs = prev_source.isna()

stitch_break = (
    final_data["stitch_block"]
    != prev_stitch_block
)

family_break = (
    (final_data["source_contrcode"] == prev_source)
    &
    (final_data["block_id"] != prev_family_block)
)

final_data["new_return_block"] = (
    first_obs
    | stitch_break
    | family_break
)

final_data["return_block"] = (
    final_data
    .groupby("paper_ticker")["new_return_block"]
    .cumsum()
    .astype(int)
)


# ------------------------------------------------------------
# Daily arithmetic return
# ------------------------------------------------------------

final_data["daily_return"] = (
    final_data
    .groupby(
        ["paper_ticker", "return_block"]
    )["continuous_price"]
    .pct_change(fill_method=None)
)

# protect against numerical nonsense
nonfinite = (
    final_data["daily_return"].notna()
    & ~np.isfinite(final_data["daily_return"])
)

final_data.loc[
    nonfinite,
    "daily_return"
] = np.nan


# ============================================================
# Wide return matrix
# ============================================================

paper64_returns = (
    final_data
    .pivot(
        index="date_",
        columns="paper_ticker",
        values="daily_return",
    )
    .sort_index()
)


# ============================================================
# Availability matrix
# ============================================================

paper64_available = (
    paper64_returns.notna()
)


# ============================================================
# Save final model-ready data
# ============================================================

final_data.to_parquet(
    DATA_DIR / "paper64_model_data_v3.parquet",
    index=False
)

paper64_returns.to_parquet(
    DATA_DIR / "paper64_returns_v3.parquet"
)

paper64_available.to_parquet(
    DATA_DIR / "paper64_availability_v3.parquet"
)


# ============================================================
# MATERIAL QA ONLY
# ============================================================

n_blocks = (
    final_data[
        ["paper_ticker", "return_block"]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "Markets:",
    final_data["paper_ticker"].nunique()
)

print(
    "Return matrix markets:",
    paper64_returns.shape[1]
)

print(
    "Return blocks:",
    n_blocks
)

print(
    "NaN returns:",
    final_data["daily_return"].isna().sum()
)

print(
    "Non-finite returns:",
    nonfinite.sum()
)

print(
    "Max absolute daily return:",
    final_data["daily_return"].abs().max()
)

print(
    "|return| > 50%:",
    (
        final_data["daily_return"].abs() > 0.50
    ).sum()
)

print("\nSaved final model-ready datasets under Data/")
```
